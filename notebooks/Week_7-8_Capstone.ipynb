## 1. Describe the people and pick a home

```python
import report, homes

occupants = [
    {'name': 'Sam', 'year': 1990, 'gender': 'female'},
    {'name': 'Lee', 'year': 1985, 'gender': 'male'},
]
home = homes.load_home('sunny_studio')
```

## 2. The full report

```python
rep = report.full_report(home, occupants)
print('Home:', rep['home'])
print('Occupants:')
for p in rep['occupants']:
    print(f"   {p['name']}: Kua {p['kua']} ({p['group']} group)")
fs = rep['flying_stars']
print('Flying Stars: Period', fs['period'],
      '(AMBIGUOUS — re-measure)' if fs['ambiguous'] else '(charted)')
```

## 3. Form School findings

```python
form = rep['form']
print('Missing corners:', form['missing_corners'] or 'none')
bed = form['bed_command']
print('Bed:', bed['reason'] if bed else 'no bed (or no door) to assess')
print('Poison arrows:', form['poison_arrows'] or 'none')
```

## 4. ✏️ YOUR CELL — the conflict table
This is the heart of the whole project. For each sector, show the compass area, the BTB area, and mark conflicts. Then add each person's Eight Mansions quality there.

```python
print(f"{'Direction':>10} | {'Compass area':>14} | {'BTB area':>14} | conflict?")
print('-' * 60)
for row in rep['conflict_table']:
    flag = 'DISAGREE' if row['conflict'] else ''
    print(f"{row['direction']:>10} | {row['compass_area']:>14} | {row['btb_area']:>14} | {flag}")
n = sum(r['conflict'] for r in rep['conflict_table'])
print(f'\n{n} sectors where the compass school and BTB disagree.')
```

## 5. ✅ The report is complete and honest

```python
checks = [
    ('all four schools present', {'flying_stars','form','conflict_table','occupants'} <= set(rep)),
    ('conflict table has 9 sectors', len(rep['conflict_table']) == 9),
    ('at least one conflict surfaced', any(r['conflict'] for r in rep['conflict_table'])),
    ('a disclaimer is attached', len(rep['disclaimer']) > 0),
]
for label, ok in checks: print(('✅' if ok else '❌'), label)
assert all(ok for _, ok in checks)
```
