## 1. Which period is a home?
Periods are 20-year blocks. We're in Period 9 (2024–2043).

```python
import flying_stars as fs, homes

for yr in [2010, 2025, 1995]:
    print(yr, '-> Period', fs.period_of(yr))
```

## 2. Build a natal chart
Each palace holds [mountain star, base star, facing star].

```python
home = homes.load_home('sunny_studio')
period = fs.period_of(home['construction_year'])
chart = fs.natal_chart(period, home['facing_degrees'])
for d, (mtn, base, face) in chart.items():
    print(f'{d:>10}: mountain {mtn}  base {base}  facing {face}')
```

## 3. Draw it (South is up, the flying-star convention)

```python
import viz
viz.plot_flying_star_chart(chart, f'{home["name"]} — Period {period}')
plt.show()
```

## 4. ✅ Reproduce published charts, cell-for-cell
Our 'accuracy': does the code match real published natal charts in `data/reference/flying_star_charts.json`?

```python
all_ok = True
for ref in homes.load_reference('flying_star_charts')['charts']:
    c = fs.natal_chart(ref['period'], ref['facing_degrees'])
    ok = all(c[d] == ref['palaces'][d] for d in ref['palaces']); all_ok &= ok
    print(('✅' if ok else '❌'), f"chart {ref['id']} reproduced exactly")
assert all_ok
```

## 5. ✏️ YOUR CELL — the honest refusal
One sample home faces a 24-mountain boundary on purpose. A real practitioner re-measures; the code refuses. Catch the `AmbiguousFacing` exception and print its message.

```python
edge = homes.load_home('edge_case_flat')
p = fs.period_of(edge['construction_year'])
try:
    fs.natal_chart(p, edge['facing_degrees'])
    print('charted (unexpected)')
except fs.AmbiguousFacing as exc:
    print('Refused, correctly:')
    print(' ', exc)
```
