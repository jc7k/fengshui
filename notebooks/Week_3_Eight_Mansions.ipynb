## 1. Your Kua number
From birth year + gender. Note the formula changed in the year 2000.

```python
import eight_mansions as em

for year, gender in [(1990,'female'), (1985,'male'), (2010,'male')]:
    k = em.kua_number(year, gender)
    print(f"{gender} born {year}: Kua {k} ({em.group(k)} group)")
```

## 2. ✅ Check against the published table
Every example in `data/reference/kua_table.json`.

```python
import homes
ref = homes.load_reference('kua_table')
all_ok = True
for ex in ref['examples']:
    m = em.kua_number(ex['year'], 'male'); f_ = em.kua_number(ex['year'], 'female')
    ok = (m == ex['male'] and f_ == ex['female']); all_ok &= ok
    print(('✅' if ok else '❌'), f"{ex['year']}: male={m} female={f_}")
assert all_ok
```

## 3. Your four good and four bad directions

```python
kua = em.kua_number(1990, 'female')
print('Kua', kua, '- GOOD directions:')
for d, q in em.good_directions(kua).items(): print(f'   {d:>10}: {q}')
print('Kua', kua, '- BAD directions:')
for d, q in em.bad_directions(kua).items(): print(f'   {d:>10}: {q}')
```

## 4. ✏️ YOUR CELL — is the bed facing a good direction?
A home's bed has a `facing`. Snap it to a direction and check its Eight Mansions quality for this person.

```python
import compass
home = homes.load_home('sunny_studio')
bed = homes.find_feature(home, 'bed')
bed_dir = compass.direction_of(bed['facing'])
quality = em.direction_quality(kua, bed_dir)
verdict = 'auspicious' if em.is_auspicious(kua, bed_dir) else 'inauspicious'
print(f'Bed faces {bed_dir}: {quality} ({verdict}) for Kua {kua}')
```

## The Li Chun honesty note
The feng shui year starts ~Feb 4, not Jan 1.

```python
print('Jan 1990, no date  ->', em.kua_number(1990, 'male'))
print('Jan 15 1990        ->', em.kua_number(1990, 'male', 1, 15), '(counts as 1989!)')
print('Mar 15 1990        ->', em.kua_number(1990, 'male', 3, 15))
```
