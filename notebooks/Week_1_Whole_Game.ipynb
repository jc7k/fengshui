## 1. Load a sample home
A home is a tiny JSON file: a grid of rooms plus a facing direction. No download — it ships in the repo.

```python
import homes, compass, analyzer, viz

print('Available homes:', homes.list_homes())
home = homes.load_home('sunny_studio')
home
```

## 2. Draw it
Row 0 is the north edge; the red square is the front door.

```python
viz.plot_home(home)
plt.show()
```

## 3. Snap the facing to a direction

```python
facing = compass.direction_of(home['facing_degrees'])
print(f"{home['name']} faces {home['facing_degrees']} deg -> {facing}")
```

## 4. Lay the bagua over the home (BTB door-aligned)
Which life area does each room fall in?

```python
for entry in analyzer.room_areas(home):
    print(f"{entry['room']:>8} -> {entry['area']}")
```

## 5. ✏️ YOUR CELL — print the naive recommendations

## 6. See the overlay
The life-area map drawn on the home.

```python
import bagua
door_wall = homes.find_feature(home, 'door')['wall']
viz.plot_bagua_overlay(home, bagua.btb_overlay(door_wall), 'BTB overlay')
plt.show()
```

## ✅ Did it work?
A quick self-check (this is our stand-in for 'accuracy').

```python
checks = [
    ('facing snapped to a word', isinstance(facing, str)),
    ('every room got a life area', len(analyzer.room_areas(home)) > 0),
    ('at least one recommendation', len(analyzer.naive_recommendations(home)) >= 1),
]
for label, ok in checks:
    print(('✅' if ok else '❌'), label)
assert all(ok for _, ok in checks)
```
