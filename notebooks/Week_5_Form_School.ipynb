## 1. Missing corners
An L-shaped home is 'missing' a bagua sector — that life area has no physical space.

```python
import floorplan, homes

for name in ['sunny_studio', 'courtyard_L']:
    h = homes.load_home(name)
    print(f"{name:>13}: missing {floorplan.missing_corners(h) or 'nothing'}")
```

## 2. ✅ Check the geometry against hand-worked answers

```python
L = homes.load_home('courtyard_L')
studio = homes.load_home('sunny_studio')
checks = [
    ('courtyard_L missing Northeast', floorplan.missing_corners(L) == ['Northeast']),
    ('studio missing nothing', floorplan.missing_corners(studio) == []),
]
for label, ok in checks: print(('✅' if ok else '❌'), label)
assert all(ok for _, ok in checks)
```

## 3. Command position
Can the bed/desk see the door without sitting in its direct line?

```python
for feat in ['bed', 'desk']:
    r = floorplan.command_position(studio, feat)
    print(f"{feat}: {'IN command' if r['in_command'] else 'not commanding'} — {r['reason']}")
```
