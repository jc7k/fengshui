## 1. The year pillar
One Heavenly Stem + one Earthly Branch per year.

```python
import bazi

for yr in [1984, 1990, 2024]:
    p = bazi.year_pillar(yr)
    print(f"{yr}: {p['stem']}-{p['branch']} — {p['polarity']} {p['element']} {p['animal']}")
```

## 2. ✅ 1984 was the start of a 60-year cycle (Jia-Zi)
A known anchor.

```python
p = bazi.year_pillar(1984)
ok = (p['stem'] == 'Jia' and p['branch'] == 'Zi')
print(('✅' if ok else '❌'), f"1984 = {p['stem']}-{p['branch']} (Wood Rat)")
assert ok
```

## 3. ✏️ YOUR CELL — your own pillar
Put in your birth year (add month/day to apply the ~Feb-4 boundary) and read your year pillar.

```python
me = bazi.year_pillar(1995, month=6, day=1)
print(f"Year pillar: {me['stem']}-{me['branch']}")
print(f"That's a {me['polarity']} {me['element']} {me['animal']} year.")
```
